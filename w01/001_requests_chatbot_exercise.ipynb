````python
from anthropic import Anthropic
client = Anthropic()
model="claude-sonnet-5"
````

````python
def add_user_message(messages, user_message):
    messages.append({"role": "user", "content": user_message})
    return messages
````

````python
def add_assistant_message(messages, assistant_message):
    messages.append({"role": "assistant", "content": assistant_message})
    return messages
````

````python
def chat(messages, stop_sequences=None):
    params = {
        "model": model,
        "messages": messages,
        "max_tokens": 300
    }
    if stop_sequences:
        params["stop_sequences"] = stop_sequences
    response = client.messages.create(**params)
    return response.content[0].text
````

````python
messages=[]
while True:
    user_input = input(">: ")
    if user_input.strip().lower() in ("quit", "exit", "q"):
        print("Goodbye!")
        break
    print(">:", user_input)
    add_user_message(messages, user_input)
    chat_response = chat(messages)
    add_assistant_message(messages, chat_response)
    print(">Assistant:", chat_response)
````

## Math tutor

````python
def math_tutor(messages,system=None):
    params={
        "model": model,
        "messages": messages,
        "max_tokens": 300
    }
    if system:
        params["system"] = system
    response = client.messages.create(**params)
    return response.content[0].text
````

````python
messages=[] 
system="You are a helpful and patient math tutor. Don't directly answer student questions. Guide them for success."
add_user_message(messages, "Explain the Pythagorean theorem to me.")
math_tutor_response = math_tutor(messages)
print(">Assistant:", math_tutor_response)
````

## Python Chatbot

````python
def python_programmer(messages, system=None):
    params={
        "model":model,
        "messages":messages,
        "max_tokens":300
    }
    if system:
        params["system"] = system

    response=client.messages.create(**params)
    return next(b.text for b in response.content if b.type == "text")  
````

````python
messages=[]
add_user_message(messages, "Write a Python function that checks for duplicates in a string.")
system="You are a skilled Python programmer. Provide clear and efficient code solutions."   
python_response=python_programmer(messages)
````

````python
print(python_response,end="\n\n")
````

## Temperature test

````python
def temp_test(messages,temp=None):
    params={
        "model":model,
        "messages":messages,
        "max_tokens":300
    }
    if temp:        
        params["temperature"] = temp   
    response=client.messages.create(**params)
    return next(b.text for b in response.content if b.type == "text")
````

## Response Streaming

````python
messages=[]
add_user_message(messages,"Explain what is a database in 2 sentences")
response=client.messages.create(
    model=model,
    messages=messages,
    max_tokens=300,
    stream=True
)
for event in response:
    print(event)
````

````python
messages=[]
add_user_message(messages, "Description of a database in 1 sentence?")

with client.messages.stream(
    model=model,
    messages=messages,
    max_tokens=300
) as stream:
    for text in stream.text_stream:
        print(text,end="")

stream.get_final_message()
````

## Structured Data

````python
def str_chat(messages, stop_sequences=None):
    params = {
        "model": "claude-sonnet-5",
        "messages": messages,
        "max_tokens": 300
    }
    if stop_sequences:
        params["stop_sequences"] = stop_sequences
    response = client.messages.create(**params)
    return next(b.text for b in response.content if b.type == "text")
````

````python
messages=[]
add_user_message(messages, "Generate a very short event bridge rule as json")
x=str_chat(messages)
print(x,end="\n\n")
````

````python
messages=[]
add_user_message(messages, "Generate a very short event bridge rule as json.")
add_assistant_message(messages, "```json")
x=str_chat(messages,stop_sequences=["```"])
print(x)
````
